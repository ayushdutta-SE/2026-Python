```python
user_input=''
start=False
 
while True:
  user_input=input('> ').lower()
  if user_input=='start':
    
    if start:
      print('Car is already Started')
      
    else:
      start=True
      print('Car Started...')
  elif user_input=='stop':
    
    if not start:
      print('Car is already at halt')
    else:
      start=False
      print('Car Stopped....')
  elif user_input=='help':
    print('''
Start-To start the car
Stop=to stop the car
Quit-To exit''')
  elif user_input=='quit':
    break
  else:
    print('Wrong input')
```
