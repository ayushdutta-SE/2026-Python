### Check if a number entered by the user is positive, negative, or zero.

```python
number =int(input('Enter the number: '))

if number<0:
  print('Number is negative')
elif number>0:
  print('Number is positive')
else:
  print('Number is zero')
```

### Take two numbers as input. Print which one is larger, or 'Equal' if both are same.

```python
num1=int(input('Number 1> '))
num2=int((input('Number 2> ')))

if num1==num2:
  print('Equal')
elif num1>num2:
  print(f'{num1} is larger than {num2}')
else:
  print(f'{num2} is larger than {num1}')
```

### Given a student's score (0–100), print the grade: A (≥90), B (≥75), C (≥60), D (≥45), F otherwise.

```python
marks=int(input('ENTER MARKS: '))

if marks>=90:
  print('GRADE A')
elif marks>=75:
  print('GRADE B')
elif marks>=60:
  print('GRADE C')
elif marks>=45:
  print('GRADE D')
else:
  print('GRADE F')
```

### Check if a year is a leap year. (Divisible by 4, but not 100, unless also divisible by 400.)

```python
year=int(input())
print(f'Year is {year}')

if year%4==0:
  if year%100==0:
    if year%400==0:
      print('Leap Year')
    else:
      print('Not')
  else:
    print('Leap year')
else:
    print('Not a Leap Year')
```

### Print all numbers from 1 to 50. But for multiples of 3 print 'Fizz', multiples of 5 print 'Buzz', multiples of both print 'FizzBuzz'.

```python
for item in range(0,51):
  if int(item)%3==0:
    if int(item)%5==0:
      print('FizzBuzz')
    else:
      print('Fizz')
  elif int(item)%5==0:
    if int(item)%3==0:
      print('FizzBuzz')
    else:
      print('Buzz')
  else:
    print(item)
```

### Print the multiplication table of a number entered by the user (from 1 to 10).

```python
number=int(input())

for i in range(1,11):
  result=1
  result=number*i
  print(f'{number} x {i} = {result}')
```

### Find the sum of all even numbers between 1 and 100 using a loop.

```python
sum=0
for item in range(1,101):
  if item%2==0:
    sum+=item
print(f'Sum> {sum} ')
```
