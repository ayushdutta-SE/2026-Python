```python
number=[34,20,56,99,98]
max=number[0]

for item in number:
  if item>max:
    max=item
print(max)
```

```python
numbers=[5,3,4,7,1,2,7,6,5]
unique=[]
numbers.sort()
print(numbers)

for num in numbers:
  if num not in unique:
    unique.append(num)
  


print(unique)
```

```python
phone=input('Phone: ')

digit_map={
  "1":'One',
  "2":'Two',
  "3":'Three',
  "4":'Four'
}
output=''
for num in phone:
  output+=digit_map.get(num,'#')+ " "
print(output)
```

```python
message=input()

new=message.split(" ")

emoji={
  ":)":"😊",
  ":(":"😒"
}
output=''
for word in new:
  output+=emoji.get(word,word)+ " "
print(output)
```
